```python
import pandas as pd
import os

folder_path = r"C:\Users\Jeen priyee\Downloads\gda_till_march2026"

# pick all Excel files
all_files = [f for f in os.listdir(folder_path) if f.lower().endswith(".xlsx")]

print("Excel files found:", len(all_files))

if not all_files:
    raise ValueError("No Excel files found in the folder.")

df_list = []

for file in all_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_excel(file_path)
    df_list.append(df)

merged_df = pd.concat(df_list, ignore_index=True)

# save as CSV
output_path = os.path.join(folder_path, "gda_updated_31-03-2026.csv")
merged_df.to_csv(output_path, index=False, encoding="utf-8")

print("Merging complete!")
print("Final rows, columns:", merged_df.shape)
```

```python
import pandas as pd

gda_updated = pd.read_csv(r"C:\Users\Jeen priyee\Downloads\gda_till_march2026\gda_updated_31-03-2026.csv")
gda_updated.head()
```

```python
gda_updated = gda_updated.sort_values(by="Compliant ID", ascending=False)
```

```python
ids_to_delete = [11111111, 2222222]

# Filter the DataFrame to exclude these IDs
gda_updated = gda_updated[~gda_updated['Compliant ID'].isin(ids_to_delete)]

# Check the updated DataFrame
print(gda_updated)
```

```python
gda_updated.loc[:, 'Compliant ID'] = pd.to_numeric(gda_updated['Compliant ID'], errors='coerce')
gda_updated.shape
```

```python
all_ids = set(range(
    int(gda_updated['Compliant ID'].min()),
    int(gda_updated['Compliant ID'].max()) + 1))

present_ids = set(gda_updated['Compliant ID'].dropna().astype(int))

missing_ids = sorted(list(all_ids - present_ids))
missing_ids
```

```python
missing_complaints = pd.DataFrame({
    "Compliant ID": [
        74993, 74994], "Status": ["Unrelated", "Unrelated"]})

gda_march2026 = pd.concat([gda_updated, missing_complaints], ignore_index=True)
```

```python
print("Total complaints:", len(gda_march2026))
gda_march2026["Status"].value_counts()
```

```python
gda_march2026.shape
```

```python
gda_march2026 = gda_march2026.drop_duplicates(subset="Compliant ID", keep="first")
gda_march2026.shape
```

```python
gda_march2026['Date and Time'] = pd.to_datetime(gda_march2026['Date and Time'],format='%Y-%m-%d %H:%M:%S',errors='coerce')

gda_march2026['Month'] = gda_march2026['Date and Time'].dt.strftime('%b %Y')

gda_march2026[['Date and Time', 'Month']].head()
```
